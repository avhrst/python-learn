```python
y = torch.tensor([5],dtype=torch.float32)
x1 = torch.tensor([3],dtype=torch.float32)
x2 = torch.tensor([1],dtype=torch.float32)

w1 = torch.tensor([0.5],requires_grad=True, dtype=torch.float32)
w2 = torch.tensor([0.5],requires_grad=True, dtype=torch.float32)

x1w1 = torch.mul(x1,w1)
x2w2 = torch.mul(x2,w2)

result = x1w1+x2w2
result

loss = torch.nn.functional.mse_loss(result,y)
loss.backward()

w1.grad

```
